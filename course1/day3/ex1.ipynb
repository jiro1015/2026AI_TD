## (2) 振り返り（提出）

＿＿ に書き込む（それぞれ 1〜3 文）．

1. ① で紹介した自分の制作物は，5 分類のどれに当たるか（AI を使っていなければ「該当なし」）．その根拠：自分の制作物は、主に「1 AIアプリケーション開発」に当たる。既存のAIサービスやAPIを利用してアプリケーションを開発するもので、AIモデル自体を一から学習させているわけではないためである。
2. ② の自分たちの答えと ③ の答え合わせで違った所．なぜ違ったか：AIを使った開発はすべて同じだと考えていたが、AIアプリケーションの開発とAIモデル自体の開発では分類が異なることが分かった。AIを利用することと、AIそのものを学習・改良することの違いを十分に理解していなかったためだと考える。
3. 演習1 全体を通して学んだこと．「AI をつくる」の見方が，演習の前と後でどう変わったか：演習前は、AIをつくることは自分でAIモデルを学習させることだと思っていた。しかし、演習を通して、既存のAIやAPIを活用してアプリケーションを開発することもAI開発の一つだと分かった。
4. 次につくってみたい AI を 1 つ挙げ，5 分類のどれか，必要な技術と計算環境，今の自分たちの PC でどこまでできるか（◎○△）を書く：掲示板の投稿から株価の変動や投資家の心理を分析するAIをつくってみたい。分類は「3-A 特化型AIモデル開発」で、Python、データ収集、機械学習、自然言語処理の技術が必要になる。一般的なPCでもデータ処理や小規模なモデルの学習は可能なので、今のPCでできる範囲は「○」だと考える。
